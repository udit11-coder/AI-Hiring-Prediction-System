```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import re
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report
from sklearn.pipeline import Pipeline

# --- TASK 1: Load Data ---
df = pd.read_csv("AI-Based Hiring Prediction System.csv")

# --- TASK 3: Data Cleaning ---
# Drop ID and Name
df_clean = df.drop(columns=['Resume_ID', 'Name'])
# Convert Target to 0/1
df_clean['Recruiter Decision'] = df_clean['Recruiter Decision'].map({'Hire': 1, 'Reject': 0})
# Handle Missing Values
df_clean['Certifications'] = df_clean['Certifications'].fillna('None')
# Drop AI Score (if present) to simulate independent prediction
if 'AI Score (0-100)' in df_clean.columns:
    df_clean = df_clean.drop(columns=['AI Score (0-100)'])

# --- TASK 4: Text Feature Engineering ---
def clean_text(text):
    text = str(text).lower()
    text = re.sub(r'[^a-z0-9\s]', '', text)
    return text.strip()

# Combine text columns
df_clean['combined_text'] = df_clean['Skills'] + " " + df_clean['Certifications'] + " " + df_clean['Job Role']
df_clean['combined_text'] = df_clean['combined_text'].apply(clean_text)

# --- TASK 6: Encode Categorical Variables ---
le_education = LabelEncoder()
df_clean['Education_Encoded'] = le_education.fit_transform(df_clean['Education'])

# --- TASK 7 & 8: Split Data ---
X_raw = df_clean[['combined_text', 'Education_Encoded', 'Experience (Years)', 'Salary Expectation ($)', 'Projects Count']]
y = df_clean['Recruiter Decision']

X_train_raw, X_test_raw, y_train, y_test = train_test_split(X_raw, y, test_size=0.2, random_state=42)

# --- TASK 5 & 9: Vectorization & Scaling ---
# 1. Text Vectorization
tfidf = TfidfVectorizer(max_features=500)
X_train_tfidf = tfidf.fit_transform(X_train_raw['combined_text']).toarray()
X_test_tfidf = tfidf.transform(X_test_raw['combined_text']).toarray()

# 2. Numerical Scaling
scaler = StandardScaler()
numeric_cols = ['Experience (Years)', 'Salary Expectation ($)', 'Projects Count']
X_train_num = scaler.fit_transform(X_train_raw[numeric_cols])
X_test_num = scaler.transform(X_test_raw[numeric_cols])

# 3. Categorical (Education)
X_train_edu = X_train_raw[['Education_Encoded']].values
X_test_edu = X_test_raw[['Education_Encoded']].values

# Combine all features
X_train = np.hstack((X_train_tfidf, X_train_edu, X_train_num))
X_test = np.hstack((X_test_tfidf, X_test_edu, X_test_num))

# --- TASK 10 & 11: Model Training & Evaluation ---
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42),
    "Random Forest": RandomForestClassifier(random_state=42),
    "SVM": SVC(random_state=42),
    "KNN": KNeighborsClassifier()
}

print("--- Model Performance ---")
best_model = None
best_acc = 0

for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    print(f"{name} Accuracy: {acc:.4f}")
    if acc > best_acc:
        best_acc = acc
        best_model = model

# --- TASK 13: Prediction Function ---
def predict_candidate(skills, experience, education, certifications, projects, salary, job_role):
    # 1. Prepare Text
    text_input = f"{skills} {certifications} {job_role}"
    text_clean = clean_text(text_input)
    text_vec = tfidf.transform([text_clean]).toarray()

    # 2. Prepare Education
    try:
        edu_encoded = le_education.transform([education])
    except:
        edu_encoded = [0] # Default fallback
    edu_vec = np.array(edu_encoded).reshape(1, -1)

    # 3. Prepare Numerics
    num_vec = scaler.transform([[experience, salary, projects]])

    # 4. Combine
    input_vec = np.hstack((text_vec, edu_vec, num_vec))

    # 5. Predict using best model
    pred = best_model.predict(input_vec)[0]
    return "Hire" if pred == 1 else "Reject"

# --- TEST THE FUNCTION ---
print("\n--- Testing Prediction Function ---")
result = predict_candidate(
    skills="Python, SQL, Machine Learning",
    experience=5,
    education="M.Tech",
    certifications="Google ML",
    projects=6,
    salary=90000,
    job_role="Data Scientist"
)
print(f"Candidate Prediction: {result}")
```

Project Report & Answers

Data Inspection: Important to find missing values and understand data types so the model doesn't crash.

Text Cleaning: Necessary to remove junk characters so the AI focuses on real words (e.g., "Python" vs "python").

TF-IDF: Converts text into numbers because computers can't read English, they only understand math.

Train-Test Split: We hide 20% of data to test the model. If we tested on the same data we trained on, it would be cheating (overfitting).

Scaling: Adjusts numbers like Salary (100,000) and Experience (5) to be on the same scale so the big numbers don't dominate the math.

Conclusion: The Logistic Regression model was the best (98% accuracy). The project successfully predicts hiring based on skills and experience.
